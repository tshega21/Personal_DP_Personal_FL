```python
import numpy as np
import ast
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict

sns.set_style('whitegrid', {'axes.linewidth': 1, 'axes.edgecolor':'black'}) 

methods = ['global', 'pooled']
method_labels = {
    'global': 'Global Model Accuracy',
    'pooled': 'Personal Model Accuracy'
}
num_steps = 10
linestyle_map = {
    'global': (0, (5, 1)),
    'pooled': (0, (5, 2)),
}

enames = ["0.1"]
datasets = ['No Privacy', 'Global Privacy', 'Local & Global Privacy']
dataset_colors = ['C0', 'C1']  # colors for global and pooled

fig, axs = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

for d_idx, ds in enumerate(datasets):
    ax = axs[d_idx]
    results = all_results[d_idx]

    # Dataset filtering
    if ds.lower() == 'no privacy':
        df = results[results['e'] == 'PrivacyFree']
    else:
        df = results[results['e'].isin(enames)]

    # Filter for number of personal steps
    df = df[df['num_personal_steps'] == num_steps]

    for m_idx, method in enumerate(methods):
        df_method = df[df['type'] == method]
        if len(df_method) == 0:
            continue

        lambda_accs = defaultdict(list)
        for _, row in df_method.iterrows():
            lam = row['lambda']
            final_acc = row['mean_perf']
            lambda_accs[lam].append(final_acc)

        # Average per lambda and sort
        lambdas_sorted = sorted(lambda_accs.keys())
        final_mean = [np.mean(lambda_accs[lam]) for lam in lambdas_sorted]

        # Plot
        ax.plot(
            lambdas_sorted, final_mean,
            label=method_labels[method],
            linestyle=linestyle_map[method],
            marker='o',
            color=dataset_colors[m_idx % 2]
        )

    ax.set_xscale('log')

    ax.set_title(ds)
    ax.set_xlabel("Lambda")
    if d_idx == 0:
        ax.set_ylabel("Final Accuracy")
    ax.set_ylim(0, 1)
    ax.grid(True)
    ax.legend()

fig.suptitle(f" Test Accuracy vs Lambda (Steps={num_steps})")
plt.tight_layout()
plt.show()
```
